# Practical 1 — Study of Deep Learning Packages

Study the main deep-learning frameworks and run small examples where the required library is available.

## Aim
Study the deep learning packages TensorFlow, Keras, PyTorch and Theano.

## Notes
This notebook checks the available package installations and demonstrates small, safe examples using TensorFlow/Keras and PyTorch. Theano is included as a theoretical/legacy package: it is discontinued and may not install or import on modern Python versions. The notebook intentionally does not fail if Theano is unavailable.

```python
# 1. Check Python and package versions
import sys
print("Python version:", sys.version)

def show_version(package_name):
    try:
        module = __import__(package_name)
        print(f"{package_name}: {getattr(module, '__version__', 'version not exposed')}")
        return module
    except Exception as exc:
        print(f"{package_name}: unavailable ({type(exc).__name__}: {exc})")
        return None

tf = show_version("tensorflow")
keras = show_version("keras")
torch = show_version("torch")
theano = show_version("theano")
```

```python
# 2. TensorFlow/Keras: build and run a tiny neural network
if tf is not None:
    from tensorflow import keras as tf_keras
    from tensorflow.keras import layers

    model = tf_keras.Sequential([
        layers.Input(shape=(4,)),
        layers.Dense(8, activation="relu"),
        layers.Dense(3, activation="softmax")
    ])
    model.compile(optimizer="adam",
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])

    # Small synthetic data: 60 examples, 4 features, 3 integer classes
    import numpy as np
    rng = np.random.default_rng(7)
    X = rng.normal(size=(60, 4)).astype("float32")
    y = rng.integers(0, 3, size=(60,))
    model.fit(X, y, epochs=2, batch_size=8, verbose=0)
    print("TensorFlow/Keras prediction:", model.predict(X[:1], verbose=0))
else:
    print("Install TensorFlow to run this section: pip install tensorflow")
```

```python
# 3. PyTorch: build a tiny neural network and run a forward pass
if torch is not None:
    import torch.nn as nn

    torch.manual_seed(7)
    pytorch_model = nn.Sequential(
        nn.Linear(4, 8),
        nn.ReLU(),
        nn.Linear(8, 3)
    )
    sample = torch.randn(1, 4)
    logits = pytorch_model(sample)
    probabilities = torch.softmax(logits, dim=1)
    print("PyTorch output probabilities:", probabilities.detach().numpy())
else:
    print("Install PyTorch to run this section: see https://pytorch.org/get-started/locally/")
```

## Result
The notebook identifies installed package versions and demonstrates a basic forward/training pass in TensorFlow/Keras and a forward pass in PyTorch. Theano availability is reported without interrupting the notebook.
